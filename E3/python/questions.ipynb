```python
def return_indices_of_given_target(list):
    target = 9
    new_list_of_indices= []
    for outer in range(len(list)):
        for inner in range(outer+1,len(list)):
            if inner > outer:
                result = list[outer] + list[inner]
                if result == target:
                    new_list_of_indices.append((outer, inner))

    print(new_list_of_indices)

nums = [2, 7,2, 11, 15]
return_indices_of_given_target(nums)            
```

```python
def return_indices_of_given_target(nums, target):
    seen = {}  # value -> index
    for i, num in enumerate(nums):
        complement = target - num
        if complement in seen:
            return (seen[complement], i)
        seen[num] = i
    return None

nums = [2, 7,2, 11, 15]
target = 9
return_indices_of_given_target(nums,target)  
```

```python
"""
Question: Write a function that checks whether a given string is a palindrome
(reads the same forwards and backwards), ignoring case and spaces.
"""
def is_palindrome(given_string):
    cleaned = given_string.replace(" ","")
    lower_cleaned_string = cleaned.lower()
    reversed_text = ""
    for char in lower_cleaned_string:
        reversed_text = char + reversed_text  
        
    if(lower_cleaned_string == reversed_text):
        result = "palindrome"
    else:
        result = "not palindrome"
        
    return result

given_string="A man a plan a canal Panama"
result = is_palindrome(given_string)
print(result)



```

```python
def is_palindrome(given_string):
    cleaned = given_string.replace(" ", "").lower()
    return cleaned == cleaned[::-1]

print(is_palindrome("A man a plan a canal Panama"))  # True
print(is_palindrome("Hello World"))   
```

```python
"""
Question: Given a list of words, write a function that counts how many times each word appears,
and returns the word(s) with the highest frequency.

words = ["apple", "banana", "apple", "orange", "banana", "apple"]
most_frequent_word(words)  
# should return "apple" (appears 3 times)
"""

def most_frequent_word(words):
    counts = {}  # dictionary to hold word -> count

    for word in words:
        if word in counts:
            counts[word] += 1
        else:
            counts[word] = 1

    # now find the word with the max count
    max_word = None
    max_count = 0
    for word, count in counts.items():
        if count > max_count:
            max_count = count
            max_word = word

    return max_word
    

words = ["apple", "banana", "apple", "orange", "banana", "apple"]
most_frequent_word(words)  

    
    
```

```python
from collections import Counter

def most_frequent_word(words):
    counts = Counter(words)
    return counts.most_common(1)[0][0]  # most_common(1) returns [("apple", 3)]

words = ["apple", "banana", "apple", "orange", "banana", "apple"]
most_frequent_word(words) 
```
