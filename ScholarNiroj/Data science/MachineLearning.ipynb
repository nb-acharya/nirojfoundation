- Supervised learning: we need labeled data. We need both feature and target
    - ****Regression:****
        - linear regression
        - Random Forest Regressor
        - Decision Tree Regressor
        - Support Vector Regressor
        - XGBoost Regressor

        ---> If our target data is continuous data like salary, weather, price......
  
    - ****Classification:****
        - Logistic Regression (only for binary classification)
        - Random Forest Classifier

        - Support Vector Classifier

##### Metrics
to check how well our machine learning model performed.
1. regression:
- RMSE(Root Mean Square Error)
- MAE(Mean Absolute Error)
- R2 score -> Regression Accuracy
- MSE (Mean Square Error)
- Rule: Regression model performs well when RMSE value is low and R2 score is high.

2. Classification
- Classification Report (precision, recall, f1-score, support, accuracy)
- Confusion Matrix

3. Clustering
- Silhouette Score

##### Feature Scaling (used in numerical columns)
- convert numeric data into range
1. standard scaler -> used when our data contains outliers
   - (Xi-mean) / sd
   - -3 to 3 range is fine
3. min max scaler -> used when our data does not contain outliers
   - (Xi-min_val) / (max_val - min_val)
   - range should be in between 0 to 1

##### Label Encoding (used in categorical columns)
- convert alphabetical data into numeric data
1. label encoder -> used for data which can be compared
2. one hot encoder -> used for data which cannot be compared and it only contains binary value(0(no) and 1(yes))
3. feature encoder
4. encryption encoder
