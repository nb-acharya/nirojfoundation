### Linear Regression

Is used to find the relationship between variables. In a ML context, linear regression finds the relationship between features and label.
y' = b+w1x1
b =  bias
w1 = weight

y' = b + w1x1+w2x2+w3x3+w4x4+w5x5

The bias and weights are updated during training.


**Loss**
Loss is a numerical metric that describes how wrong the model's predictions are. It is a distance between model prediction and the actual labels.
Loss is measured from the actual value to the predicted value.

**Five main types of loss**
L1 loss
Mean Absolute Error(MAE)
L2 loss
Mean squared error(MSE)
Root Mean Squared Error(RMSE)
