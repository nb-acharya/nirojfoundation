### Hypothesis testing

```python
#20260610
import pandas as pd
from scipy import stats
```

### One sample t-test

```python
df=pd.read_csv("pizza_sales.csv")
df.head()
```

```python
df.info()
```

###### Is the average total_price of all pizza sales different from $15

```python
sample_mean = 15
pop = df['total_price']
alpha = 0.05
```

```python
t_statistics,p_value=stats.ttest_1samp(pop,popmean = sample_mean)
```

```python
## formula:
#(actual_mean - sample_mean) / (sd / root(total_number_of_data))
import numpy as np
actual_mean = df['total_price'].mean()
sd = df['total_price'].std()
root = np.sqrt(48620)
first_part = actual_mean - sample_mean
second_part = sd / root
final_answer = first_part / second_part
final_answer
```

```python
Ho = 'Ho: Average total price of all pizza sales are equal to $15.'
Ha = 'H1: Average total price of all pizza sales is not equal to $15'
print(Ho)
print(Ha)
print(f'Statistics: {t_statistics:,.2f}')
print(f'Probability Value: {p_value:,.2f}')
print(f'Actual Average Price: Rs.{df["total_price"].mean():,.2f}')
print(f'Sample Average Price: Rs. {sample_mean}')
```

```python
if p_value < alpha:
    print(f'Decision: Failed to accept null hypothesis.\nConclusion: Needs to accept alternate hypothesis\nAnswer: {Ha}')
else:
    print(f'Decision: Failed to reject null hypothesis.|\nConclusion:  Need to accept null hypothesis.\nAnswer:{Ho}')
```
