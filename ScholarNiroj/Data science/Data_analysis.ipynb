```python
import pandas as pd
```

```python
df= pd.read_csv("Employees.csv")
df.head()
```

```python
df['COMMISSION_PCT'] = df["COMMISSION_PCT"].fillna(0.22)
df["MANAGER_ID"] = df["MANAGER_ID"].fillna(100.0)
df["DEPARTMENT_ID"] = df["DEPARTMENT_ID"].fillna(50.0)
```

```python
df["MANAGER_ID"] = df["MANAGER_ID"].astype(int)
df["DEPARTMENT_ID"] = df["DEPARTMENT_ID"].astype(int)
```

```python
df['HIRE_DATE'] = pd.to_datetime(df['HIRE_DATE'])
```

```python
df.info()
```

```python
'''
groupby()
aggregate function -> mean() max() min() count() sum()

If we use groupby clause in our analysis then aggregate function must be used
If we only use aggregate function then group by clause is not compulsory to use
'''
```

```python
# Find total employees working in each departments
emp_dep = df.groupby('DEPARTMENT_ID')['EMPLOYEE_ID'].count().reset_index()
emp_dep
```
