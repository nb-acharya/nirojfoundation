```python
#20260506
import pandas as pd
```

```python
df = pd.read_csv("Employees.csv")
df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
commission_mean = round(df['COMMISSION_PCT'].mean(),2)
commission_mean
```

```python
df['COMMISSION_PCT'].median()
```

```python
df["COMMISSION_PCT"].fillna(commission_mean,inplace= True)
```

```python
### advised to use mode for the categorical column if they have any null value to be filled.
```

```python
manager_mode = df['MANAGER_ID'].mode()[0]
manager_mode
```

```python
department_mode = df['DEPARTMENT_ID'].mode()[0]
department_mode
```

```python
df['MANAGER_ID'] = df['MANAGER_ID'].fillna(manager_mode)
df['DEPARTMENT_ID'] = df['DEPARTMENT_ID'].fillna(department_mode)
```

```python
df['DEPARTMENT_ID'].isnull().sum()
```

```python
'''
still there is null, because if we have multiple values with same frequency, panda will  not decide on 
which mode it will use, so we have to use the 0th index and specify that
'''
```
