```python
import pandas as pd
```

```python
df = pd.read_excel("EMPLOYEES.xlsx")
```

```python
df.head()
```

```python
df.isnull().sum().sum()
```

#### conditional filtering is same as boolean indexing in numpy

```python
## find details of employees whose salaary is more than 18000
df[df['SALARY'] > 18000]
```

```python
## find details of employees whose salary is in the range of 10000 and 20000
sal_range = (df["SALARY"] >= 10000) & (df["SALARY"] <= 20000)
df[sal_range]
```

```python
sal_between = df["SALARY"].between(10000,20000)
sal_between
```

```python
#  OR used for caategorical record
# find details of employees who works in job: SA_REP, IT_PROG
job_details = (df["JOB_ID"] == "SA_REP") | (df["JOB_ID"] == "IT_PROG")
df[job_details]
```

```python
job_in = df["JOB_ID"].isin(["SA_REP", "IT_PROG"])
df[job_in]
```

```python
# find details of employees whose salary is in range of 6000 to 10000 and they must be working on job SA_REP
job_filter = (df['SALARY'] >=6000) & (df["SALARY"] <=10000) & (df["JOB_ID"] == "SA_REP")  
df[job_filter]
```
